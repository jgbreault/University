# Pipeline 10 Rule Extraction Walkthrough

This notebook is a presentation/demo companion for `official_rule_extraction_pipeline`.

It is intentionally **cache-first**: the cells below read existing Pipeline 10 artifacts and preview each step. They do not call Gemini or rerun the expensive extraction by default.

Change `CITY` in the setup cell to inspect `burnaby`, `calgary`, or `vancouver`.

## 0. Setup

Load small helper functions and point the notebook at one city output directory.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Markdown

CITY = "calgary"  # try: "burnaby", "calgary", "vancouver"

ROOT = Path.cwd()
PIPELINE = ROOT / "code" / "official_rule_extraction_pipeline"
CITY_DIR = PIPELINE / "outputs" / CITY
HQ_DIR = CITY_DIR / "06_rule_extraction_hq"

def read_json(path):
    return json.loads(Path(path).read_text(encoding="utf-8"))

def read_jsonl(path, limit=None):
    rows = []
    with Path(path).open("r", encoding="utf-8") as handle:
        for line in handle:
            if line.strip():
                rows.append(json.loads(line))
                if limit and len(rows) >= limit:
                    break
    return rows

def df_jsonl(path, limit=10):
    return pd.DataFrame(read_jsonl(path, limit=limit))

def compact_dict(d):
    return pd.DataFrame([{"key": k, "value": v} for k, v in d.items()])

print(f"CITY={CITY}")
print(f"CITY_DIR={CITY_DIR}")
print(f"HQ_DIR exists: {HQ_DIR.exists()}")
```

## 1. Local Candidate Selection

Input: city PDF under `pdfs/<city>/`.

Output: locally scored text blocks and candidate windows. This stage is deterministic and does not call an API.

```python
local_dir = CITY_DIR / "01_local_selection"
local_summary = read_json(local_dir / "summary.json")
display(compact_dict({
    "source_pdf": local_summary.get("source_pdf"),
    "pdf_page_count": local_summary.get("pdf_page_count"),
    "local_block_count": local_summary.get("local_block_count"),
    "candidate_window_count": local_summary.get("candidate_window_count"),
    "recommended_visual_page_count": local_summary.get("recommended_visual_page_count"),
}))

blocks = df_jsonl(local_dir / "local_blocks_scored.jsonl", limit=8)
cols = [c for c in ["block_id", "page_number", "selection_score", "target_hits", "parent_hits", "universal_hits", "text"] if c in blocks.columns]
display(blocks[cols])
```

## 2. Auto Discovery

This stage expands candidate windows into discovered regulatory blocks, assigns rough lanes, and records why each block was selected.

```python
discovery_dir = CITY_DIR / "02_auto_discovery"
discovery_summary = read_json(discovery_dir / "summary.json")
display(compact_dict({
    "selected_block_count": discovery_summary.get("selected_block_count"),
    "target_hit_block_count": discovery_summary.get("target_hit_block_count"),
    "low_confidence_selected_count": discovery_summary.get("low_confidence_selected_count"),
}))
display(pd.DataFrame([{"lane": k, "count": v} for k, v in discovery_summary.get("lane_counts", {}).items()]))

disc = df_jsonl(discovery_dir / "discovered_blocks.jsonl", limit=8)
cols = [c for c in ["block_id", "page_number", "selected_because", "target_hits_discovered", "rule_types", "text"] if c in disc.columns]
display(disc[cols])
```

## 3. Semantic Compression

This stage reduces the discovered block set before extraction. The default provider is lexical; optionally, local Ollama embeddings can be used.

```python
compression_dir = CITY_DIR / "03_semantic_compression"
compression_summary = read_json(compression_dir / "summary.json")
display(compact_dict({
    "provider": compression_summary.get("provider"),
    "threshold": compression_summary.get("threshold"),
    "input_selected_block_count": compression_summary.get("input_selected_block_count"),
    "compressed_block_count": compression_summary.get("compressed_block_count"),
    "dropped_block_count": compression_summary.get("dropped_block_count"),
    "compression_ratio": compression_summary.get("compression_ratio"),
}))

compressed = df_jsonl(compression_dir / "compressed_blocks.jsonl", limit=8)
cols = [c for c in ["block_id", "page_number", "compression_lane", "semantic_score", "structural_score", "compression_score", "text"] if c in compressed.columns]
display(compressed[cols])
```

## 4. Graph/RAG Pack Builder

Compressed blocks are grouped into RAG packs by applicability lane: core rules, use permissions, district dimensional overrides, and universal applicable rules.

```python
rag_dir = CITY_DIR / "04_graph_rag_packs"
rag_summary = read_json(rag_dir / "experiment_summary.json")
display(compact_dict({
    "selected_block_count": rag_summary.get("selected_block_count"),
    "evidence_block_count": rag_summary.get("evidence_block_count"),
    "context_pack_count": rag_summary.get("context_pack_count"),
    "duplicate_cluster_count": rag_summary.get("duplicate_cluster_count"),
}))

pack_files = sorted(rag_dir.glob("*_packs.jsonl"))
display(pd.DataFrame([{"file": p.name, "pack_count": len(read_jsonl(p))} for p in pack_files]))

sample_pack_file = rag_dir / "district_dimensional_overrides_packs.jsonl"
sample_pack = read_jsonl(sample_pack_file, limit=1)[0]
display(pd.DataFrame([{k: sample_pack.get(k) for k in ["pack_id", "city", "page_number", "scope", "applicability"]}]))
display(pd.DataFrame(sample_pack.get("evidence_blocks", [])[:3]))
```

## 5. RAG Visual-Block Adapter

The adapter converts graph/RAG packs into the visual-block format consumed by the extractor. It also applies target trimming and inherits bundled table regions/crops when available.

```python
visual_dir = CITY_DIR / "05_rag_visual_blocks"
adapter_manifest = read_json(visual_dir / "rag_adapter_manifest.json")
display(compact_dict({
    "pack_count": adapter_manifest.get("pack_count"),
    "text_block_count": adapter_manifest.get("text_block_count"),
    "target_filter_enabled": adapter_manifest.get("target_filter_enabled"),
    "copied_table_region_count": adapter_manifest.get("copied_table_region_count"),
    "copied_table_image_count": adapter_manifest.get("copied_table_image_count"),
}))
display(pd.DataFrame([{"action": k, "count": v} for k, v in adapter_manifest.get("target_filter_actions", {}).items()]))

text_blocks = df_jsonl(visual_dir / "text_blocks.jsonl", limit=8)
cols = [c for c in ["block_id", "original_page_number", "rag_lane", "target_filter_action", "text"] if c in text_blocks.columns]
display(text_blocks[cols])

table_regions_path = visual_dir / "table_regions.jsonl"
table_regions = read_jsonl(table_regions_path) if table_regions_path.exists() and table_regions_path.stat().st_size else []
display(pd.DataFrame(table_regions[:5]) if table_regions else Markdown("No table regions for this city/run."))
```

## 6. Extraction Batch Planning

The extractor batches text blocks by RAG pack density and character budget. Table regions are handled separately as image batches.

```python
batch_manifest_path = HQ_DIR / "batch_manifest.jsonl"
if batch_manifest_path.exists():
    batches = df_jsonl(batch_manifest_path, limit=20)
    cols = [c for c in ["batch_id", "lanes", "rag_pack_ids", "original_pages", "source_count", "char_count"] if c in batches.columns]
    display(batches[cols])
else:
    display(Markdown("No HQ batch manifest yet. Run extraction or dry-run extraction first."))

preflight_path = HQ_DIR / "preflight_skipped_sources.json"
if preflight_path.exists():
    preflight = read_json(preflight_path)
    display(pd.DataFrame(preflight[:10]) if preflight else Markdown("No preflight-skipped sources."))
```

## 7. LLM Extraction Results

This stage reads cached parsed results from `api_raw/` when available. Text batches and table-image batches are merged into raw rules.

```python
summary_path = HQ_DIR / "summary.json"
hq_summary = read_json(summary_path)
display(compact_dict({
    "text_successful_batches": hq_summary.get("text_successful_batches"),
    "text_error_batches": hq_summary.get("text_error_batches"),
    "table_successful_batches": hq_summary.get("table_successful_batches"),
    "text_rule_count": hq_summary.get("text_rule_count"),
    "table_rule_count": hq_summary.get("table_rule_count"),
    "combined_rule_count": hq_summary.get("combined_rule_count"),
}))

text_logs = read_json(HQ_DIR / "text_extraction_logs.json") if (HQ_DIR / "text_extraction_logs.json").exists() else []
display(pd.DataFrame(text_logs)[["batch_id", "status", "source_count", "rule_count", "skipped_count", "seconds"]] if text_logs else Markdown("No text logs."))

table_logs = read_json(HQ_DIR / "table_extraction_logs.json") if (HQ_DIR / "table_extraction_logs.json").exists() else []
display(pd.DataFrame(table_logs) if table_logs else Markdown("No table image extraction logs."))
```

## 8. Merge, Dedup, Clean Final, and Review Queue

`merged_rules_deduplicated` is the filtered/deduplicated full set. `merged_rules_clean` is the default downstream output. `merge_review_queue` keeps incomplete or conflicted rules for manual review.

```python
display(compact_dict({
    "deduplicated_rule_count": hq_summary.get("deduplicated_rule_count"),
    "clean_rule_count": hq_summary.get("clean_rule_count"),
    "merge_review_rule_count": hq_summary.get("merge_review_rule_count"),
    "postprocess_filtered_rule_count": hq_summary.get("postprocess_filtered_rule_count"),
}))
display(pd.DataFrame([{"reason": k, "count": v} for k, v in hq_summary.get("extraction_quality_flags", {}).items()]))

clean_rules = pd.read_csv(HQ_DIR / "merged_rules_clean.csv")
display(clean_rules.groupby("rule_object").size().sort_values(ascending=False).reset_index(name="count").head(12))
display(clean_rules.head(10))

review_path = HQ_DIR / "merge_review_queue.csv"
review_rules = pd.read_csv(review_path) if review_path.exists() else pd.DataFrame()
display(review_rules.head(10) if len(review_rules) else Markdown("Review queue is empty."))
```

## 9. Table Extraction Preview

If a city has regulatory table crops, table-image rules appear in `table_rules_raw.*`. If there are no table regions, this stage is empty.

```python
table_rules_path = HQ_DIR / "table_rules_raw.csv"
table_rules = pd.read_csv(table_rules_path) if table_rules_path.exists() else pd.DataFrame()
if len(table_rules):
    display(table_rules.head(12))
    display(table_rules.groupby("source_id").size().reset_index(name="rule_count"))
else:
    display(Markdown("No table-image rules in this city/run."))
```

## 11. Takeaways

- Pipeline 10 is self-contained under `code/official_rule_extraction_pipeline`.
- The expensive API call layer is separated from deterministic PDF selection, compression, and pack building.
- `merged_rules_clean.*` is the demo-friendly final output.
- `merge_review_queue.*` is useful for showing why some rules are held back from downstream GIS processing.
