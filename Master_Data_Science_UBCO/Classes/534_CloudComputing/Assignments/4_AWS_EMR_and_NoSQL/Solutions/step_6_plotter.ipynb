```python
import matplotlib.pyplot as plt
import pandas as pd
import glob
import os

path = './ngram_results'
all_files = glob.glob(os.path.join(path, "part-*"))

dataframes = []
for filename in all_files:
    df = pd.read_csv(filename, sep='\t', names=['Year', 'Count'])
    dataframes.append(df)

full_df = pd.concat(dataframes).sort_values('Year')

plt.figure(figsize=(10, 6))
plt.plot(full_df['Year'], full_df['Count'], marker='o', color='blue', linestyle='-')

plt.title('Occurrences of "Google" in Ngrams by Year')
plt.xlabel('Year')
plt.ylabel('Total Count')
plt.grid(True)

plt.savefig('Step6.png')
plt.show()
```
